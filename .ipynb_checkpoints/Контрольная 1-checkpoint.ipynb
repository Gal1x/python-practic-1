# 1. Программа сложения двух чисел

````python
a = float(input("Введите первое число: "))
b = float(input("Введите второе число: "))

result = a + b

print("Сумма:", result)
````

## Блок-схема

```mermaid
flowchart TD
    A([Начало]) --> B[/Ввод a/]
    B --> C[/Ввод b/]
    C --> D["result = a + b"]
    D --> E[/Вывод result/]
    E --> F([Конец])

# 2. Программа деления двух чисел

````python
a = float(input("Введите первое число: "))
b = float(input("Введите второе число: "))

if b != 0:
    result = a / b
    print("Результат деления:", result)
else:
    print("На ноль делить нельзя!")
````

## Блок-схема

```mermaid
flowchart TD
    A([Начало]) --> B[/Ввод a/]
    B --> C[/Ввод b/]
    C --> D{b != 0?}
    D -- Да --> E["result = a / b"]
    E --> F[/Вывод result/]
    D -- Нет --> G[/Вывод сообщения об ошибке/]
    F --> H([Конец])
    G --> H

# 3. Программа возведения числа в степень

````python
n = float(input("Введите основание степени: "))
m = float(input("Введите показатель степени: "))

result = n ** m

print("Результат:", result)
````

## Блок-схема

```mermaid
flowchart TD
    A([Начало]) --> B[/Ввод n/]
    B --> C[/Ввод m/]
    C --> D["result = n ** m"]
    D --> E[/Вывод result/]
    E --> F([Конец])
